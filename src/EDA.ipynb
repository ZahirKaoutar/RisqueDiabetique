```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("../data/dataset-diabete.csv")
print(df.head())
print(df.shape)
print(df.isna().sum())
print(df.duplicated().sum())
display(df.describe())
```

```python
df = df.drop(columns=["Unnamed: 0"])
```

```python
num_cols = df.select_dtypes(include="number").columns

for c in num_cols:
    plt.figure(figsize=(8, 4))

    sns.histplot(df[c], kde=True, bins=10)

    plt.axvline(
        df[c].mean(),
        color='black',
        linestyle='--',
        label=f"Moyenne : {df[c].mean():.2f}"
    )

    plt.axvline(
        df[c].median(),
        linestyle='-',
        color='red',
        label=f"Médiane : {df[c].median():.2f}"
    )

    plt.title(f"Distribution de {c}")
    plt.xlabel(c)
    plt.ylabel("Nombre d'observations")
    plt.legend()
    plt.show()
```
